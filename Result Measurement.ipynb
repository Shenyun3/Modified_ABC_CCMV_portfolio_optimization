```python
def difference(ucef, ccef):
    """
    Compute horizontal (risk) and vertical (return) percentage deviation errors
    relative to the interpolated unconstrained efficient frontier (Chang et al., 2000).
    Each entry in ucef and ccef is [return (y), risk (x)].
    """
    x_diff_set = []
    y_diff_set = []

    # Sorted by return y (col 0) and by risk x (col 1)
    sorted_by_y = np.array(sorted(ucef, key=lambda p: p[0]), dtype=float)
    sorted_by_x = np.array(sorted(ucef, key=lambda p: p[1]), dtype=float)
    y_vals = sorted_by_y[:, 0]
    x_vals = sorted_by_x[:, 1]

    for y_star, x_star in ccef:
        # Interpolate x** on UCEF at return y_star
        idx_y = np.searchsorted(y_vals, y_star, side='right')
        if 0 < idx_y < len(sorted_by_y):
            yk, xk_alt = sorted_by_y[idx_y - 1]
            yj, xj_alt = sorted_by_y[idx_y]
            if yj != yk:
                x_ss = xk_alt + (xj_alt - xk_alt) * (y_star - yk) / (yj - yk)
            else:
                x_ss = xk_alt
        elif idx_y == 0:
            x_ss = sorted_by_y[0, 1]
        else:
            x_ss = sorted_by_y[-1, 1]

        # Interpolate y** on UCEF at risk x_star
        idx_x = np.searchsorted(x_vals, x_star, side='right')
        if 0 < idx_x < len(sorted_by_x):
            yk_alt, xk = sorted_by_x[idx_x - 1]
            yj_alt, xj = sorted_by_x[idx_x]
            if xj != xk:
                y_ss = yk_alt + (yj_alt - yk_alt) * (x_star - xk) / (xj - xk)
            else:
                y_ss = yk_alt
        elif idx_x == 0:
            y_ss = sorted_by_x[0, 0]
        else:
            y_ss = sorted_by_x[-1, 0]

        x_diff = abs(100.0 * (x_star - x_ss) / x_ss) if x_ss != 0 else np.inf
        y_diff = abs(100.0 * (y_star - y_ss) / y_ss) if y_ss != 0 else np.inf

        x_diff_set.append(x_diff)
        y_diff_set.append(y_diff)

    return x_diff_set, y_diff_set
```

```python
def measurement(x_diff, y_diff):
    """
    Calculate percentage error summary metrics (Chang et al., 2000).
    x_diff: risk (standard deviation) percentage deviations
    y_diff: return percentage deviations
    """
    pe = np.minimum(x_diff, y_diff)
    MEAPE = float(np.mean(pe))
    MEDPE = float(np.median(pe))
    MINPE = float(np.min(pe))
    MAXPE = float(np.max(pe))

    return MEAPE, MEDPE, MINPE, MAXPE
```

```python
def show(ef):
    ef = np.array(sorted(ef, key=lambda x: x[0]))
    sum_return = sum(ef[:,0])
    sum_risk = sum(ef[:,1])
    mean_return = np.mean(ef[:,0])
    mean_risk = np.mean(ef[:,1])
    median_return = np.median(ef[:,0])
    median_risk = np.median(ef[:,1])
    return sum_return,sum_risk,mean_return,mean_risk,median_return,median_risk
```
